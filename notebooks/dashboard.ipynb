# InferArena Dashboard

This notebook explores experiment results produced by the ``inferarena`` CLI.
Run an experiment first:

```bash
inferarena compare --config examples/experiment.yaml --schedulers fcfs,chunked_prefill,priority,sjf
```

Then restart and run all cells in this notebook.

````python
import matplotlib.pyplot as plt

from inferarena.reporting.dashboard import (
    discover_experiments,
    plot_latency_cdf,
    to_dataframe,
)

%matplotlib inline
````

## Discover experiments

````python
summaries = discover_experiments("../inferarena_outputs")
print(f"Discovered {len(summaries)} experiment(s)")
df = to_dataframe(summaries)
df.head()
````

## Summary table

````python
summary_cols = [
    "scheduler",
    "completed_requests",
    "throughput_rps",
    "ttft_p50_ms",
    "ttft_p99_ms",
    "latency_p50_ms",
    "latency_p99_ms",
    "cache_hit_rate",
]
display_cols = [c for c in summary_cols if c in df.columns]
df[display_cols].sort_values("throughput_rps", ascending=False)
````

## Latency CDF comparison

````python
fig, ax = plt.subplots(figsize=(8, 5))
plot_latency_cdf(summaries, ax=ax)
plt.tight_layout()
plt.show()
````
